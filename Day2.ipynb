```python
import random
print("WELCOME")
secretno=random.randint(1,100)
limit=5
print("YOU WILL BE HAVE",limit,"ATTEMPTS TO GUESS THE NUMBER")
for i in range(1,limit+1):
  guess=int(input("Enter your guess"))
  if guess==secretno:
    print("Correct")
    break
  elif guess>secretno:
    print("Too High!!!!")
  else:
    print("Too Low!!!!")
else:
  print("GAME OVER!!!")
  print("TRY AGAIN")
  print("The correct number was:", secretno)
```
